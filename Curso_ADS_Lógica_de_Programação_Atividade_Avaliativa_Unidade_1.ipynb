```python
notas = [] #Esta linha determina a lista 'notas' como vazia
#para receber as notas


while len(notas) < 3: #Esta linha diz que: enquanto houve menos de 3 variáveis
#na lista notas, deve fazer o comando a seguir

#Aqui pedimos ao aluno para indicar o valor da variável a ser inserida em notas
  nota = float(input('Digite a nota do trimestre: (De 0 a 10)'))

#O comando a baixa determina que: Se o valor digitado pelo usuário estiver
#entre 0 e 10, sera inserido dentro da lista, caso não esteja, sera mostrado que
# o valor deve estar entre 0 e 10.
  if 0 <= nota <= 10:
    notas.append(nota)
  else:
    print('A nota deve estar entre 0 e 10')

media = sum(notas)/len(notas) #Aqui é calculado o valor da media, dividindo
#a soma das notas pela quantidade inserida.

#Abaixo, temos a impressão das notas inseridas e as condições para o aluno ser
#aprovado ou reprovado.
print(f'\nAs suas notas foram: {notas}')

if media > 6:
  print(f"A media do aluno foi: {media:.2f}.\nAluno Aprovado")
else:
  print(f'A media do aluno foi: {media:.2f}.\nAluno Reprovado')
```
